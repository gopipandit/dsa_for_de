***Question 1- Given a list of integers, write a function to find the longest increasing subsequence.***

```python
def longest_increasing_subsequence(arr):
    if not arr:
        return []
    
    n = len(arr)
    
    dp = [1] * n 
    for i in range(1, n):
        for j in range(i):
            if arr[i] > arr[j]:
                dp[i] = max(dp[i], dp[j] + 1)

    return max(dp)
```

```python
nums = [10, 9, 2, 5, 3, 7, 101, 18]

result = longest_increasing_subsequence(nums)

print(result)
```

***Question-2 Write a function to find the top N frequent elements in a list.***



```python
def get_counts(nums):
    for i in set(nums):
        print(i, "->",nums.count(i))
```

```python
nums = [1, 1, 1, 2, 2, 3]
```

```python
get_counts(nums)
```

```python
def top_n_frequent(nums,n):
    freq = {}
    #Count of occurences
    for i in nums:
        freq[i] = freq.get(i, 0) + 1
        
    sorted_items = sorted(freq.items(), key=lambda x: x[1], reverse=True)
    
    return [item[0] for item in sorted_items[:n]]
```

```python
print(top_n_frequent(nums, 2))
```

***Question 3- Write a function to calculate the moving average of a list of numbers.***

```python
def get_moving_average(nums, n):
    moving_average = []
    if len(nums) < n:
        return moving_average
    for i in range(0,len(nums)-(n-1)):
        moving_average.append(sum(nums[i:i + n]) / n)
    return moving_average
```

```python
nums = [10, 20, 30, 40, 50,45,34,56,67,87]
```

```python
get_moving_average(nums,4)
```

***Question 4 - Write a Python function to calculate the median of a list of numbers.***

```python
def get_median(nums):
    n = len(nums)
    nums = sorted(nums)
    if n % 2 == 0:
        result = (nums[n // 2 - 1] + nums[n // 2])/2
    else:
        result = nums[n // 2]
        
    return result
```

***Question 5- Write a Python function to find the intersection of two lists.***

```python
def get_intersection(lst1, lst2):
    result = []
    for i in lst1:
        if i in lst2:
            result.append(i)
    return result
```

```python
lst1 = [1, 2, 3, 4]
lst2 = [3, 4, 5, 6]

print(get_intersection(lst1, lst2))
```

```python
def get_intersection(lst1, lst2):
    return list(set(lst1).intersection(set(lst2)))
```

```python
lst1 = [1, 2, 3, 4]
lst2 = [3, 4, 5, 6]

print(get_intersection(lst1, lst2))
```

***Question 6 -Write a Python function that takes a list of integers and returns a list of integers that are the result of applying the square root to each element, ignoring any negative numbers.***

```python
def get_sqrt(num):
    return [i**1/2 for i in num if i > 0]
```
