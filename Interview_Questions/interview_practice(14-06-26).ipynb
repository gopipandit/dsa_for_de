***Question 2- Write a function that takes a list of strings and returns a dictionary with the frequency of each string.***

```python
words = ["apple", "banana", "apple","apple" ,"orange", "banana", "apple"]

def get_frequecy(l):
    result_dict = {i: l.count(i) for i in set(l)}
    sorted_result = sorted(result_dict, key = lambda x:result_dict[x], reverse = True)
    return sorted_result
```

```python
get_frequecy(words)
```

```python
get_frequecy(words)
```

***Question 4- Write a function to calculate the moving average of a list of numbers.***

```python
nums = [10, 20, 30, 40, 50,45,34,56,67,87]
```

```python
def get_moving_avg(nums,n):
    result = []
    for i in range(0, len(nums) - (n - 1)):
        result.append(sum(nums[i:i + n])/n)
    return result
    
    
```

```python
get_moving_avg(nums,3)
```

```python
n = 3
for i in range(0,len(nums)-(n-1)):
    print(nums[i:i+n])
```

***question 5 -Write a Python function to flatten a nested dictionary.***

```python
nums = [3,4,5,3,5,[67,89,78,[56,78]],12,34]
```

```python
def flatten_list(nums):
    result = []
    
    for i in nums:
        if isinstance(i, list):
            result.extend(flatten_list(i))
        else:
            result.append(i)
    return result
```

```python
flatten_list(nums)
```

***Question 6- Write a Python function to calculate the factorial of a number using recursion.***

```python
def get_factorial(n):
    if n < 0:
        raise ValueError("n is negative")
    if n == 1:
        return 1
    else:
        return n * get_factorial(n-1)
```

```python
get_factorial(8)
```

***Question 7- Write a function that takes a list of integers and returns the list sorted in ascending order without using built-in sort functions.***

```python
def get_sorted_list(nums):
    
    for i in range(len(nums)):
        for j in range(i+1,len(nums)):
            if nums[i] > nums[j]:
                nums[i], nums[j] = nums[j], nums[i]
    return nums
            
```

```python
get_sorted_list(nums2)
```
