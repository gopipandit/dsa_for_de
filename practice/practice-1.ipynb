*Problem Statement*: Given a list of integers, write a function to find and return the
maximum element in the array.
- The array will contain at least one element.
- Do not use the built-in max() function.

```python
from typing import List
```

```python
def get_max(num : List[int]):
    result = num[0]
    
    for value in num:
        if value > result:
            result = value
    return result
        
```

```python
num1 = [3, 1, 7, 2, 5]
num2 = [-10, -3, -1, -20]
```

```python
get_max(num1)
```

Problem Statement: Write a function that takes a string as input and returns the string
reversed.
- The input string may contain spaces, digits, and special characters.
- Do not use Python’s built-in [::-1] slicing.

```python
def reverse_string(text : str) -> str:
    result = ""
    for i in range(len(text),0,-1):
        result += text[i-1]     
    return result
        
    
```

```python
txt = "hello"
```

```python
reverse_string(txt)
```
