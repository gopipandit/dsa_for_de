```python
import datetime
class Human:
    def __init__(self, name, gender = None, friend_list = None, yob = None):
        self.name = name
        self.gender = gender
        self.friend_list = friend_list if friend_list else []
        self.lifespan = 70
        self.yob = yob
        

    def add_friend(self, human):
        self.friend_list.append(human)
        human.friend_list.append(self)
        print(f"{self.name} and {human.name} are now friends!!!")
        
    def __str__(self):
        return f"{self.name} - {self.gender}"
    
    def __repr__(self):
        return f"Name: {self.name} | Gender: {self.gender}"
    
    def eat(self):
        print("Hello I am eating!")
    
    def walk(self):
        print("Hello I am walking!")
        
    @classmethod    
    def change_lifespan(cls, new_lifespan):
        cls.lifespan = new_lifespan
        
    def calculate_age(yob):
        current_year = datetime.datetime.now().year
        return current_year - yob
        
    
        
```

```python
gopi = Human("Gopi", "male")
pinki = Human("Pinki", "female")
aahana = Human("Aaahana", "female")
```

***Coding Assignment: Data Type Conversion and Validation Write a Python function called `validate_and_convert` that takes a single string as input. This function should attempt to convert the input string to an integer, then to a float. If both conversions are successful, print the original string, the integer value, and the float value. If either conversion fails, catch the exception and print an appropriate error message indicating which conversion failed. For example, if the input is "3.14", the output should be: Original: 3.14, Integer: 3, Float: 3.14. If the input is "abc", the output should indicate that both integer and float conversions failed.***

```python
def validate_and_convert(text):
    try:
        float_value = float(text)
    except ValueError:
        print(f"Original: {text} → Both integer and float conversions failed")
        return

    try:
        int_value = int(float_value)
    except ValueError:
        print(f"Original: {text} → Integer conversion failed")
        return

    print(f"Original: {text}, Integer: {int_value}, Float: {float_value}")
```

```python
txt = "gpi"
```

```python
validate_and_convert(txt)
```
